```python
# drop the missing value rows  
df.dropna(inplace =True)
```

```python
duplicate_count = df.duplicated().sum()
print(f"\n Total Duplicate Rows: {duplicate_count}") # check if dataset contain duplicate value or not
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
```

```python
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format = '%d-%m-%Y')
```

```python
df.dtypes
```

```python
df['Sales'].describe().round(2)
```

```python
df['Order Year'] = (df['Order Date']).dt.year
df['Order Month'] = (df['Order Date']).dt.month
```

```python
# Save the clean Data 
df.to_csv("superstore_clean_data.csv",index = False)
print("Cleaned data saved 'superstore_clean_data.csv")
```

```python
df.columns.tolist()
```

```python
total_sales = df['Sales'].sum()

print(f"Total sales in Rupees: ₹ {total_sales.round(2)}")
```

```python
total_sales_in_lakhs = total_sales / 100000
print(f"Total Sales in Lakhs: ₹{total_sales_in_lakhs:,.2f} Lakhs")
```

```python
total_sales_in_crores = total_sales / 10000000
print(f"Total Sales in Crores: ₹{total_sales_in_crores:,.2f} Crores")
```

```python
# inserting the Data into Postgresql
from sqlalchemy import create_engine

# 1. Connect to PostgreSQL
# Format: postgresql://username:password@host:port/database_name
engine = create_engine('postgresql://postgres:Surya9370@localhost:5432/superstore_db')

# 2. Load the CSV file from your Jupyter Lab directory
df = pd.read_csv('superstore_clean_data.csv')

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')

# 3. Write data to PostgreSQL
# 'if_exists' can be 'fail', 'replace' (drops/recreates table), or 'append' (adds to existing)
df.to_sql('superstore', con=engine, if_exists='append', index=False)

print("Data inserted successfully!")
```
